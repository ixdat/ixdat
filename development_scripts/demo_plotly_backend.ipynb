# Matplotlib and Plotly

The same EC measurement uses both ixdat backends below.

```python
from base64 import b64encode
from io import BytesIO
from pathlib import Path

import matplotlib.pyplot as plt
from IPython.display import HTML

from ixdat import Measurement

data_file = next(
    path
    for path in [
        Path("test_data/biologic/Pt_poly_cv_CUT.mpt"),
        Path("../test_data/biologic/Pt_poly_cv_CUT.mpt"),
    ]
    if path.exists()
)
measurement = Measurement.read(data_file, reader="biologic")
```

```python
matplotlib_figure = measurement.plot()[0].figure
plotly_figure = measurement.plot(backend="plotly")
image = BytesIO()
matplotlib_figure.savefig(image, format="png", dpi=125, bbox_inches="tight")
matplotlib_image = b64encode(image.getvalue()).decode("ascii")
plotly_html = plotly_figure.to_html(
    full_html=False,
    include_plotlyjs="cdn",
    config={"responsive": True},
)
plt.close(matplotlib_figure)

HTML(f'''<style>
.backend-pair {{display:grid;grid-template-columns:1fr 1fr;gap:1rem}}
.backend-pair img, .backend-pair .plotly-graph-div {{max-width:100%;width:100% !important}}
</style>
<div class="backend-pair">
<div><h3>Matplotlib</h3><img src="data:image/png;base64,{matplotlib_image}"></div>
<div><h3>Plotly</h3>{plotly_html}</div>
</div>''')
```
